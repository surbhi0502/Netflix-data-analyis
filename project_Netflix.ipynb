```python
df.drop_duplicates(inplace = True)
```

```python
df.replace(
    ['N/A','NULL',' ',''],np.nan,inplace=True
)
```

```python
df.isnull().sum()
```

```python
df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
df['title'] = df['title'].str.title()
df['type'] = df['type'].str.title()
df['age_certification'] = df['age_certification'].str.title()
df['genres'] = df['genres'].str.title()
df['production_countries'] = df['production_countries'].str.title()

```

```python
df['title'].value_counts()
```

```python
numeric_cols = [
    'release_year',
    'runtime',
    'seasons',
    'imdb_score',
    'imdb_votes',
    'tmdb_popularity',
    'tmdb_score'
]
```

```python
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df.describe()
```

```python
df['runtime'] = pd.to_datetime(df['runtime'], errors='coerce')
```

```python
df["genres"] = df["genres"].fillna("Unknown")
```

```python
df["description"] = df["genres"].fillna("description")
```

```python
df["type"] = df["type"].fillna("Unknown")
```

```python
df["content_age"] = 2026-df["release_year"]
```

```python
df["is_movie"] = (df["type"] == "MOVIE").astype(int)
```

```python
df.head()
```

```python
def watch_category(score):
    if score >= 8:
        return "Highly Rated"
    elif score >= 7:
        return "Good"
    elif score >= 6:
        return "Average"
    else:
        return "Low Rated"

df["watch_category"] = df["imdb_score"].apply(watch_category)
```
